```python
# Core libraries
import pandas as pd
import sqlite3
import json

# Display settings (optional but useful)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 20)

print("Environment ready ✅")
```

```python
import os

os.listdir()
```

```python
orders = pd.read_csv("orders.csv")

orders.head()
```

```python
orders.info()
```

```python
users = pd.read_json("users.json")

users.head()
```

```python
users.info()
```

```python
# Create SQLite database in memory
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Read SQL file
with open("restaurants.sql", "r") as file:
    sql_script = file.read()

# Execute SQL script
cursor.executescript(sql_script)

print("SQL executed successfully ✅")
```

```python
restaurants = pd.read_sql("SELECT * FROM restaurants", conn)
restaurants.head()
```

```python
restaurants.info()
```

```python
orders['user_id'].isna().sum(), orders['restaurant_id'].isna().sum()
```

```python
users['user_id'].duplicated().sum()
restaurants['restaurant_id'].duplicated().sum()
```

```python
orders_users = orders.merge(
    users,
    on="user_id",
    how="left"
)
```

```python
final_df = orders_users.merge(
    restaurants,
    on="restaurant_id",
    how="left"
)
```

```python
final_df.shape
orders.shape
```

```python
final_df.columns
```

```python
final_df.to_csv("final_food_delivery_dataset.csv", index=False)

print("Final dataset created 🎉")
```

```python
final_df[final_df['membership'] == 'Gold'] \
    .groupby('city')['total_amount'] \
    .sum() \
    .sort_values(ascending=False)
```

```python
final_df.groupby('cuisine')['total_amount'] \
    .mean() \
    .sort_values(ascending=False)
```

```python
high_value_users = (
    final_df.groupby('user_id')['total_amount']
    .sum()
    .reset_index()
)

high_value_users[high_value_users['total_amount'] > 1000]['user_id'].nunique()
```

```python
bins = [3.0, 3.5, 4.0, 4.5, 5.0]
labels = ['3.0–3.5','3.6–4.0','4.1–4.5','4.6–5.0']

final_df['rating_range'] = pd.cut(final_df['rating'], bins=bins, labels=labels)

final_df.groupby('rating_range')['total_amount'] \
    .sum() \
    .sort_values(ascending=False)
```

```python
final_df[final_df['membership'] == 'Gold'] \
    .groupby('city')['total_amount'] \
    .mean() \
    .sort_values(ascending=False)
```

```python
restaurant_counts = final_df.groupby('cuisine')['restaurant_id'].nunique()
```

```python
revenue_by_cuisine = final_df.groupby('cuisine')['total_amount'].sum()
```

```python
pd.concat(
    [restaurant_counts, revenue_by_cuisine],
    axis=1,
    keys=['restaurant_count', 'total_revenue']
).sort_values('restaurant_count')
```

```python
gold_orders = final_df[final_df['membership'] == 'Gold']['order_id'].nunique()
total_orders = final_df['order_id'].nunique()

round((gold_orders / total_orders) * 100)
```

```python
restaurant_stats = (
    final_df.groupby('restaurant_name_x')
    .agg(
        avg_order_value=('total_amount', 'mean'),
        order_count=('order_id', 'count')
    )
    .reset_index()
)

restaurant_stats[
    restaurant_stats['order_count'] < 20
].sort_values('avg_order_value', ascending=False)
```
